## Python Basics Variable

```python
# 1. Declare two variables, `x` and `y`, and assign them integer values.
#  Swap the values of these variables without using any temporary variable.

x = 5
y = 10
print("Before swapping: x =", x, ", y =", y)
x, y = y, x
print("After swapping: x =", x, ", y =", y)
```

```python
# 2. Create a program that calculates the area of a rectangle. Take the length and
#width as inputs from the user and store them in variables. Calculate and
#display the area.

length = float(input("Enter the length of the rectangle: "))
width = float(input("Enter the width of the rectangle: "))
print("Length:", length)
print("Width:", width)
area = length * width
print("The area of the rectangle is:", area)
```

```python
#3. Write a Python program that converts temperatures from Celsius to
#Fahrenheit. Take the temperature in Celsius as input, store it in a variable,
#convert it to Fahrenheit, and display the result.
celsius = float(input("Enter the temperature in Celsius: "))
print("Temperature in Celsius:", celsius)
fahrenheit = (celsius * 9/5) + 32
print(f"{celsius}°C is equal to {fahrenheit}°F")
```

## TOPIC: String Based Questions

```python
# 1. Write a Python program that takes a string as input and prints the length of
# the string.
string = input("Enter a string: ")
print("You entered:", string)
print("The length of the string is:", len(string))
```

```python
# 2. Create a program that takes a sentence from the user and counts the number
# of vowels (a, e, i, o, u) in the string.
sentence = input("Enter a sentence: ")
print("You entered:", sentence)
vowels = "aeiouAEIOU"
count = 0
for char in sentence:
    if char in vowels:
        count += 1
print("The number of vowels in the sentence is:", count)
```

```python
# 3. Given a string, reverse the order of characters using string slicing and print
# the reversed string.
string = input("Enter a string: ")
print("You entered:", string)
reversed_string = string[::-1]
print("The reversed string is:", reversed_string)
```

```python
# 4. Write a program that takes a string as input and checks if it is a palindrome
# (reads the same forwards and backwards).

string = input("Enter a string: ")
print("You entered:", string)
# Remove spaces and convert to lowercase for accurate palindrome checking
cleaned_string = string.replace(" ", "").lower()
if cleaned_string == cleaned_string[::-1]:
    print("The string is a palindrome.")
else:
    print("The string is not a palindrome.")
```

```python
# 5. Create a program that takes a string as input and removes all the spaces from
# it. Print the modified string without spaces.

string = input("Enter a string: ")
print("You entered:", string)
modified_string = string.replace(" ", "")
print("The string without spaces is:", modified_string)
```

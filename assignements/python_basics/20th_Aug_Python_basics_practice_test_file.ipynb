```python
# 41. Check if two strings are anagrams.

#what is an anagram? 
# An anagram is a word or phrase formed by rearranging the letters of a different word or phrase, 
# typically using all the original letters exactly once. 
# For example, "listen" and "silent" are anagrams of each other.

string1 = input("Enter the first string: ")
string2 = input("Enter the second string: ")

# Remove spaces and convert to lowercase
string1 = string1.replace(" ", "").lower()
string2 = string2.replace(" ", "").lower()

# Check if the sorted characters of both strings are equal
is_anagram = sorted(string1) == sorted(string2)
print("Are the strings anagrams?", is_anagram)
```

```python
# 42. Encode a string using a Caesar cipher.
## what is a Caesar cipher?
# A Caesar cipher is a simple encryption technique where each letter in the plaintext is shifted a certain number of places down or up the alphabet. 
# For example, with a shift of 3, 'A' becomes 'D', 'B' becomes 'E', etc.
input_string = input("Enter a string: ")
shift = int(input("Enter the shift value (1-25): "))
print("Input string:", input_string)
print("Shift value:", shift)
encoded_string = ""
for char in input_string:
    if char.isalpha():
        shift_base = ord('A') if char.isupper() else ord('a')
        encoded_char = chr((ord(char) - shift_base + shift) % 26 + shift_base)
        encoded_string += encoded_char
    else:
        encoded_string += char
print("Encoded string:", encoded_string)
```

```python
# 43. Decode a Caesar cipher encoded string.
input_string = input("Enter the encoded string: ")
shift = int(input("Enter the shift value (1-25): "))
print("Encoded string:", input_string)
print("Shift value:", shift)
decoded_string = ""
for char in input_string:
    if char.isalpha():
        shift_base = ord('A') if char.isupper() else ord('a')
        decoded_char = chr((ord(char) - shift_base - shift) % 26 + shift_base)
        decoded_string += decoded_char
    else:
        decoded_string += char
print("Decoded string:", decoded_string)
```

```python
# 44. Find the most frequent word in a string.
# example: "the cat and the dog" -> "the" (frequency: 2)
input_string = input("Enter a string: ")
words = input_string.split()
word_count = {}
for word in words:
    word_count[word] = word_count.get(word, 0) + 1
most_frequent_word = max(word_count, key=word_count.get)
print("Most frequent word:", most_frequent_word)
print("Frequency:", word_count[most_frequent_word])
```

```python
# 47. Check if a string contains any special characters.
input_string = input("Enter a string: ")
print("Input string:", input_string)    
special_characters = "!@#$%^&*()-+"
contains_special = False
for char in input_string:
    if char in special_characters:
        contains_special = True
        break
print("Contains special characters:", contains_special)
```

```python
# 48. Remove the nth word from a string.'
## example: "the cat and the dog", n=3 -> "the cat the dog"
input_string = input("Enter a string: ")
print("Input string:", input_string)
n = int(input("Enter the position of the word to remove: "))
print("Position to remove:", n)
words = input_string.split()
if 1 <= n <= len(words):
    words.pop(n - 1)
    new_string = " ".join(words)
    print("String after removing the nth word:", new_string)
else:
    print("Invalid position. Please enter a valid position.")
```

```python
# 49. Insert a word at the nth position in a string.
input_string = input("Enter a string: ")
print("Input string:", input_string)
n = int(input("Enter the position to insert the word: "))

print("Position to insert:", n)
word_to_insert = input("Enter the word to insert: ")
print("Word to insert:", word_to_insert)
words = input_string.split()
if 1 <= n <= len(words) + 1:
    words.insert(n - 1, word_to_insert)
    new_string = " ".join(words)
    print("String after inserting the word:", new_string)
else:
    print("Invalid position. Please enter a valid position.")
```

```python
# 50. Convert a CSV string to a list of lists
# example: "a,b,c\nd,e,f\ng,h,i" -> [["a", "b", "c"], ["d", "e", "f"], ["g", "h", "i"]]
input_string =  "a,b,c\nd,e,f\ng,h,i"
print("Input CSV string:", input_string)
rows = input_string.split("\n")
print("Rows:", rows)
csv_list = []
for row in rows:
    csv_list.append(row.split(","))
print("List of lists:", csv_list)
```
